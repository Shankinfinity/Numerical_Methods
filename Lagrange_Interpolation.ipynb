### Lagrange Interpolation

**Lagrange Interpolation** constructs a single polynomial of degree \\(n-1\\) that passes exactly through \\(n\\) known data points \\((x_0,y_0), \dots, (x_{n-1}, y_{n-1})\\):
$$
P(x) = \sum_{i=0}^{n-1} y_i \, L_i(x), \qquad L_i(x) = \prod_{\substack{j=0 \\ j \neq i}}^{n-1} \frac{x - x_j}{x_i - x_j}
$$

Unlike linear interpolation, Lagrange interpolation uses **all** data points simultaneously to build a smooth polynomial fit, which can capture curvature in the data but may **oscillate** badly for many points (Runge's phenomenon).


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return np.sin(x)

x_data = np.linspace(0, np.pi, 6)
y_data = f(x_data)
```

```python
def lagrange_interpolate_table(x_data, y_data, x_query):
    n = len(x_data)
    result = 0.0

    print(f"{'i':>2} {'L_i(x_query)':>16} {'y_i':>10} {'term':>12}")
    print("-" * 50)

    for i in range(n):
        term = y_data[i]
        Li = 1.0
        for j in range(n):
            if j != i:
                Li *= (x_query - x_data[j]) / (x_data[i] - x_data[j])
        term = y_data[i] * Li
        result += term
        print(f"{i:2d} {Li:16.6f} {y_data[i]:10.6f} {term:12.6f}")

    print(f"\nP(x_query) = {result:.6f}")
    return result
```

```python
x_query = 1.0
y_interp = lagrange_interpolate_table(x_data, y_data, x_query)
print(f"True value f({x_query}) = {f(x_query):.6f}")
```

```python
plt.figure(figsize=(12, 7), dpi=120)

# Axes lines (Desmos-style)
plt.axhline(0)
plt.axvline(0)

x = np.linspace(0, np.pi, 500)
y_true = f(x)

# Build the full Lagrange polynomial over the plotting range
def lagrange_poly(x_data, y_data, xv):
    n = len(x_data)
    yv = np.zeros_like(xv)
    for i in range(n):
        Li = np.ones_like(xv)
        for j in range(n):
            if j != i:
                Li *= (xv - x_data[j]) / (x_data[i] - x_data[j])
        yv += y_data[i] * Li
    return yv

y_poly = lagrange_poly(x_data, y_data, x)

plt.plot(x, y_true, linewidth=2, label=r"$f(x)=\sin(x)$")
plt.plot(x, y_poly, linewidth=2, linestyle="--", label="Lagrange polynomial")
plt.plot(x_data, y_data, 'o', markersize=8, label="Known data points")
plt.scatter([x_query], [y_interp], s=80, zorder=5, color="red", label="Interpolated point")

plt.grid(True, alpha=0.4)
plt.xlabel("x", fontsize=12)
plt.ylabel("f(x)", fontsize=12)
plt.title("Lagrange Interpolation", fontsize=14)

plt.legend(fontsize=11)
plt.show()
```
