### Simpson's Rule

**Simpson's Rule** approximates a definite integral by fitting parabolas through groups of three consecutive points, instead of straight lines as in the trapezoidal rule:
$$
\int_a^b f(x)\,dx \approx \frac{h}{3}\left[f(x_0) + 4\sum_{i \text{ odd}}f(x_i) + 2\sum_{i \text{ even}}f(x_i) + f(x_n)\right], \qquad h=\frac{b-a}{n}
$$

Simpson's rule requires an **even** number of subintervals \\(n\\), and is exact for polynomials up to degree 3, with error \\(O(h^4)\\) — generally far more accurate than the trapezoidal rule for the same \\(n\\).


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return np.sin(x)
```

```python
def simpsons_table(f, a, b, n_values):
    print(f"{'n':>4} {'h':>12} {'integral':>14}")
    print("-" * 35)

    results = []
    for n in n_values:
        if n % 2 != 0:
            raise ValueError("n must be even for Simpson's rule")

        h = (b - a) / n
        x = np.linspace(a, b, n + 1)
        y = f(x)

        integral = y[0] + y[-1]
        integral += 4 * np.sum(y[1:-1:2])
        integral += 2 * np.sum(y[2:-1:2])
        integral *= h / 3

        results.append(integral)
        print(f"{n:4d} {h:12.6f} {integral:14.6f}")

    return results
```

```python
a, b = 0, np.pi
n_values = [2, 4, 8, 16, 32]
results = simpsons_table(f, a, b, n_values)

integral_final = results[-1]
print(f"\nSimpson's estimate (n={n_values[-1]}) = {integral_final:.6f}")
print(f"Exact value of integral = {-(np.cos(b) - np.cos(a)):.6f}")
```

```python
plt.figure(figsize=(12, 7), dpi=120)

# Axes lines (Desmos-style)
plt.axhline(0)
plt.axvline(0)

n = 8
h = (b - a) / n
x_nodes = np.linspace(a, b, n + 1)
y_nodes = f(x_nodes)

x = np.linspace(a - 0.2, b + 0.2, 500)
y = f(x)

plt.plot(x, y, linewidth=2, label=r"$f(x)=\sin(x)$")
plt.fill_between(x_nodes, y_nodes, alpha=0.3, label="Approximated region")
plt.plot(x_nodes, y_nodes, 'o-', markersize=6, label="Nodes")

plt.grid(True, alpha=0.4)
plt.xlabel("x", fontsize=12)
plt.ylabel("f(x)", fontsize=12)
plt.title("Simpson's Rule Integration", fontsize=14)

plt.legend(fontsize=11)
plt.show()
```
