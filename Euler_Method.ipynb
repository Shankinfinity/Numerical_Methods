### Euler Method

The **Euler Method** is the simplest numerical technique for solving an initial value problem
$$
\frac{dy}{dx} = f(x,y), \qquad y(x_0) = y_0
$$
It steps forward using the slope at the current point:
$$
y_{n+1} = y_n + h\,f(x_n, y_n), \qquad x_{n+1} = x_n + h
$$

The Euler method is **simple** and easy to implement, but it has only first-order accuracy (\\(O(h)\\) local error), so small step sizes \\(h\\) are needed for good accuracy.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return y - x**2 + 1

def y_exact(x):
    return (x + 1)**2 - 0.5 * np.exp(x)
```

```python
def euler_table(f, x0, y0, h, n_steps):
    print(f"{'n':>2} {'x':>10} {'y':>12} {'y_exact':>12} {'error':>12}")
    print("-" * 55)

    xs, ys = [x0], [y0]
    x, y = x0, y0

    print(f"{0:2d} {x:10.6f} {y:12.6f} {y_exact(x):12.6f} {abs(y - y_exact(x)):12.6e}")

    for n in range(1, n_steps + 1):
        y = y + h * f(x, y)
        x = x + h
        xs.append(x)
        ys.append(y)
        print(f"{n:2d} {x:10.6f} {y:12.6f} {y_exact(x):12.6f} {abs(y - y_exact(x)):12.6e}")

    return np.array(xs), np.array(ys)
```

```python
x0, y0 = 0, 0.5
h = 0.2
n_steps = 10

xs, ys = euler_table(f, x0, y0, h, n_steps)
print(f"\ny({xs[-1]:.2f}) approx = {ys[-1]:.6f}, exact = {y_exact(xs[-1]):.6f}")
```

```python
plt.figure(figsize=(12, 7), dpi=120)

# Axes lines (Desmos-style)
plt.axhline(0)
plt.axvline(0)

x_fine = np.linspace(x0, xs[-1], 500)
y_fine = y_exact(x_fine)

plt.plot(x_fine, y_fine, linewidth=2, label="Exact solution")
plt.plot(xs, ys, 'o-', markersize=6, label="Euler approximation")

plt.grid(True, alpha=0.4)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.title("Euler Method for Solving an ODE", fontsize=14)

plt.legend(fontsize=11)
plt.show()
```
